```python
# 1ro Instalá la librería:

# ◯ Instalación previa (desde la terminal):
# ----------------------------------------
# pip install yt-dlp pandas openpyxl beautifulsoup4 requests

# ◯ Paso 1: Importar las librerías necesarias
# -------------------------------------------

# Para extraer información de YouTube sin necesidad de navegador
from yt_dlp import YoutubeDL

# Para manipular y exportar datos en tablas
import pandas as pd

# Para hacer requests HTTP si quisieras scrapear otras webs (no necesario para YouTube con yt-dlp)
import requests

# Para parsear HTML en caso de scraping tradicional (no necesario para YouTube con yt-dlp)
from bs4 import BeautifulSoup
```

```python
# URL de la playlist
playlist_url = "https://www.youtube.com/playlist?list=PL2ZWqWbnARIPaJBEbGefIGbEFczuitZQc"

# Configuración de yt-dlp: metadatos sin descargar videos
ydl_opts = {
    "quiet": True,
    "skip_download": True
}

# Extraer datos completos
with YoutubeDL(ydl_opts) as ydl:
    info = ydl.extract_info(playlist_url, download=False)

videos = info.get("entries", [])
data = []

# Procesar cada video
for video in videos:
    upload_date = video.get("upload_date")
    formatted_date = pd.to_datetime(upload_date, format="%Y%m%d") if upload_date else None

    # Extraer categoría segura (puede ser lista vacía)
    categories = video.get("categories", [])
    categoria = categories[0] if categories else None

    data.append({
        "Título": video.get("title"),
        "URL": video.get("webpage_url"),
        "Descripción": video.get("description"),
        "Fecha": formatted_date,
        "Duración (seg)": video.get("duration"),
        "Vistas": video.get("view_count"),
        "Canal": video.get("uploader"),
        "Thumbnail": video.get("thumbnail"),
        "Licencia": video.get("license"),
        "Categoría": categoria
    })

# Crear DataFrame
df = pd.DataFrame(data)

# Exportar a CSV y Excel
df.to_csv("playlist_videos_full.csv", index=False, encoding="utf-8-sig")
df.to_excel("playlist_videos_full.xlsx", index=False, engine="openpyxl")

print("✅ Exportado como 'playlist_videos_full.csv' y 'playlist_videos_full.xlsx'")
```
