** COMO HACEMOS PARA GUARDAR UN ARCHIVO DE TEXTO CON PYTHON? **  
** Pregunta de entrevista para Data Scientist Positions **  
Es importante saber que python tiene funciones para hacer guardado de archivos, binarios y de texto. Como?  

```python
if response: # Si response is true
    with open("../data/raw/adult.csv","wb") as file:
        file.write(response.content)
        
    # with open es la funcion para guardado. Entre parentesis ponemos la ruta donde guardar el 
    # archivo y el nombre que le daremos.
    # luego de la coma, le indicamos el metodo de escritura o apertura. (wb = write binary)
    # En este caso es escritura.
    # luego le decimos el alias
    # Es decir, con la funcion with open podemos abrir un archivo para lo que uno necesite, ya sea
    # lectura o escritura. 
    # El with nos asegura de que el archivo una vez que se abre, se cierre. Si pongo solo el open, 
    # luego tendremos que hacer el close. 
    # Veremos que dentro de la carpeta data / raw se abra generado el archivo csv (coma separeted values).
    
    # "w": Escritura (write mode), abre el archivo para escribir en el. Si el archivo ya existe, se sobrescribira. Si no existe, se creara.
    # "b": Modo Binario (binary mode), se puede combinar con otros modos ("rb, "wb", "ab") para trabajar con archivos binarios.
```

```python
# Volveremos a trabajar con la url de wiki, ya que hay compañeros que se les cuelga la otra web.
url = "https://en.wikipedia.org/wiki/Gun_death_and_violence_in_the_United_States_by_state"
response = requests.get(url)

# Una vez que tenemos la respuesta, vamos a querer pasar la tabla a un pandas dataframe.
# O sea, traeremos la respuesta con formato. 
if response:
       soup = BeautifulSoup(response.text, "html.parser") # Aca le digo que me quiero traer la respuesta enformato texto para formatearlo como html
soup
```

```python
# Jupyter permite imprimir en pantalla hasta un limite de lineas.
# Si quiere ir a la tabla, le damos a ver como scrollable element. 
# Es importante darle una mirada, para asegurarnos de que la tabla se extrae. 
# Ya que muchas veces no se logra, por haber miles de formas de traer tablas de internet. 
```

```python
# le pediremos a soup que busque todo lo que sean tablas y me traiga la estructura.
table = soup.find("table") 
table
```

```python
# ESTRUCTURA DE LAS TABLAS EN EL CODIGO:
# <th> = headers de la tabla
# <tr> = Rows de la tabla 
# <Caption> = Titulo de la tabla
# <td> = Celda de la tabla
# Por ahora vemos que la info es legible. 
# Tener en cuenta que todo lo que trae lo interpreta en formato texto, 
# asi que los numeros como 9% los tendremos que convertir a su formato Correcto.
```

```python
# CUANTAS TABLAS TRAJO?
# La pregunta del millon es si se trajo las dos tablas o se trajo solo una. 
# Para esto vamos al output y le damos a View as a scrollable element. 
# Y nos fijamos hasta que valor extrajo, en este caso 33.3. 
# Vemos entonces que se trajo solo la primer tabla. 
# Para efectos del ejercicio nos alcanza. Pero hay que saber que la maquina 
# no es tan lista como para traerse todas las tablas. Para ello tendriamos que trabajar mas aun.
```

```python
# GENERAREMOS UNA LISTA, CON LAS CABECERAS Y LOS DATOS.
# Ahora, Procederemos a traernos la tabla y los datos, utilizando una lista.
# En este lista, nos traeremos las cabeceras y los datos.

data = []
headers = []
for th in table.find_all("th"): # Busque todas las cabeceras de la tabla
    headers.append(th.text.strip()) # Guardalas en la lista headers, convirtiendolos a texto, y strip le quita los espacios en blanco adelante y atras
headers # Verificamos que se este logrando el cometido, imprimiendo la lista de headers.
```

```python
# Continuamos con la lista de datos
for row in table.find_all("tr"): # Buscamos todas las rows, y una vez alli, 
    cells = row.find_all("td") # buscamos los datos de las celdas
    cells = [cell.text.strip() for cell in cells] # hacemos un list comprehension, donde, para cada celda dentro de las celdas, me traigo el valor, y lo limprio quitandole espacios en blanco adelante y atras, y transformandolo en texto
    data.append(cells) # Guardamos en la lista data, los valores de las celdas.
data # Verificamos que se este logrando el cometido, imprimeiendo la lista de cells.     
```

```python
# DEBUGGING
# Si el output fuera incorrecto, ejemplo, a un compañero solo le salieron parentesis rectos,
# Lo ideal para identificar el problema es ir corriendo el codigo linea a linea, e imprimir
# el resultado para entender si se esta logrando lo que se necesita en cada linea.
```

```python
# PASARLO A UN DATAFRAME.
# Ahora deberiamos pasarlo a un pandas dataframe, 
# Pero antes debemos solucionar algunos problemas, para que lo permita.
#   Problema 1: 
#       deberia tener tantas filas como columnas esperadas.
#       Como la primer fila esta vacia, genera un error. Asi que debemos eliminarla.
#   Problema 2:
#       debemos chequear que existan la misma cantidad de headers que de columnas, en este caso 6.
#       para esto, nos fijamos la primer y la ultima fila, cuantos datos tiene. Y como aca son 6, estamos ok.
#   Conclusion: Para que funcione en pandas, solo debemos eliminar la primer fila.
```

```python
# ELIMINAMOS PRIMER FILA:
data_copy = data.copy() # Hacemos una copia de la lista data, para no modificar el raw data.
del data_copy[0] # Eliminamos de alli el primer registro.
data_copy # Verificamos que se este logrando el cometido, haciendo una impresion del resultado.     
```

```python
# GENERAMOS EL PANDAS DATAFRAME:
df = pd.DataFrame(data=data_copy, columns=headers)
df
```

```python
# Ahora con el metodo info, revisamos nulls, y tipo de datos:
df.info()
```

```python
# Vemos que el tipo de datos de cada columna es object.
# Asi que tenemos que convertir los numeros a numeros. 
# y el otro problema que tendremos para resolver, es identificar todas las celdas que 
# contienen parentesis recto, ejemplo, primera fila ultima columna, para eliminarlos,
# ya que nos impiden que ese valor se identifique como numero.
# Por ultimo, a la columna de porcentajes, debemos borrarle el simbolo "%" y dividir el valor 
# entre 100 para que lo interprete como debe ser.
```
